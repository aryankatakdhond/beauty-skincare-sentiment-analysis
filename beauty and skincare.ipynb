```python

# Step 2: Investigate mismatched reviews

# Select reviews where star ratings and predicted sentiment disagree
mismatched_reviews = sample_reviews[
    ((sample_reviews["rating"] <= 2) &
     (sample_reviews["sentiment"] == "Positive")) |
    ((sample_reviews["rating"] >= 4) &
     (sample_reviews["sentiment"] == "Negative"))
].copy()

print("Total mismatched reviews:", len(mismatched_reviews))

# Display a sample of mismatched reviews
display(
    mismatched_reviews[
        ["rating", "review_title", "review_text",
         "sentiment_score", "sentiment"]
    ].head(10)
)
```

```python

# Identify rating and sentiment disagreement

rating_sentiment_mismatch = (
    ((sample_reviews["rating"] <= 2) &
     (sample_reviews["sentiment"] == "Positive")) |
    ((sample_reviews["rating"] >= 4) &
     (sample_reviews["sentiment"] == "Negative"))
)

mismatch_count = rating_sentiment_mismatch.sum()
total_reviews = len(sample_reviews)
mismatch_percentage = (mismatch_count / total_reviews) * 100

print("Total reviews:", total_reviews)
print("Mismatched reviews:", mismatch_count)
print(f"Mismatch percentage: {mismatch_percentage:.2f}%")
```

```python

# Compare customer sentiment across brands

brand_sentiment = (
    sample_reviews
    .groupby("brand_name")
    .agg(
        total_reviews=("sentiment", "count"),
        positive_reviews=(
            "sentiment",
            lambda x: (x == "Positive").sum()
        ),
        negative_reviews=(
            "sentiment",
            lambda x: (x == "Negative").sum()
        )
    )
)

brand_sentiment["positive_percentage"] = (
    brand_sentiment["positive_reviews"]
    / brand_sentiment["total_reviews"] * 100
).round(1)

brand_sentiment["negative_percentage"] = (
    brand_sentiment["negative_reviews"]
    / brand_sentiment["total_reviews"] * 100
).round(1)

# Focus on brands with at least 10 reviews
brand_sentiment = brand_sentiment[
    brand_sentiment["total_reviews"] >= 10
]

# Display brands with the highest negative sentiment percentage
brand_sentiment = brand_sentiment.sort_values(
    "negative_percentage",
    ascending=False
)

display(brand_sentiment.head(10))
```

```python

# Step 4: Explore common words in negative reviews

from collections import Counter
import re

# Select reviews classified as negative
negative_reviews = sample_reviews.loc[
    sample_reviews["sentiment"] == "Negative",
    "review_text"
].dropna()

# Combine review text
text = " ".join(negative_reviews.astype(str)).lower()

# Extract words
words = re.findall(r"\b[a-z]{3,}\b", text)

# Remove common words that add little meaning
stop_words = {
    "the", "and", "for", "was", "but", "this", "that",
    "with", "not", "have", "had", "are", "you", "its",
    "very", "just", "from", "they", "them", "too", "all",
    "product", "would", "did", "use", "used", "out", "get",
    "has", "been", "about", "after", "when", "one", "can"
}

filtered_words = [
    word for word in words if word not in stop_words
]

# Display the 20 most frequent words
word_counts = Counter(filtered_words)

print("Total negative reviews:", len(negative_reviews))
print("\nTop 20 frequently occurring words:")

for word, count in word_counts.most_common(20):
    print(f"{word}: {count}")
```

```python

# Visualize the 15 most frequent words in negative reviews

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

top_words = word_counts.most_common(15)

word_df = pd.DataFrame(
    top_words,
    columns=["Word", "Frequency"]
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=word_df,
    x="Frequency",
    y="Word"
)

plt.title("Top 15 Words in Reviews Classified as Negative")
plt.xlabel("Word Frequency")
plt.ylabel("Word")

plt.tight_layout()
plt.savefig(
    "top_negative_review_words.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python

# Find negative reviews containing selected keywords

keywords = ["dry", "skin", "work", "cream", "irritation"]

pattern = "|".join(
    rf"\b{re.escape(word)}\b" for word in keywords
)

complaint_examples = sample_reviews.loc[
    (sample_reviews["sentiment"] == "Negative") &
    sample_reviews["review_text"].str.contains(
        pattern,
        case=False,
        na=False,
        regex=True
    ),
    ["rating", "brand_name", "product_name",
     "review_text", "sentiment_score"]
]

print("Matching reviews:", len(complaint_examples))

display(complaint_examples.head(10))
```

```python

# Count keyword-based complaint themes among negative reviews

negative_text = (
    sample_reviews.loc[
        sample_reviews["sentiment"] == "Negative",
        "review_text"
    ]
    .fillna("")
    .str.lower()
)

themes = {
    "Product effectiveness": [
        "did not work", "doesn't work", "no results",
        "not work", "didn't work", "not effective"
    ],
    "Dryness and hydration": [
        "dry", "dryness", "hydration", "moisture"
    ],
    "Irritation and discomfort": [
        "irritation", "irritating", "redness",
        "burning", "itching", "raw"
    ],
    "Texture and application": [
        "pilling", "pills", "greasy", "sticky",
        "gritty", "thick", "texture"
    ],
    "Fragrance": [
        "scent", "fragrance", "smell", "scented"
    ]
}

theme_results = []

for theme, keywords in themes.items():
    pattern = "|".join(
        re.escape(keyword) for keyword in keywords
    )

    count = negative_text.str.contains(
        pattern, regex=True, na=False
    ).sum()

    theme_results.append({
        "Theme": theme,
        "Matching reviews": int(count),
        "Percentage of negative reviews": round(
            count / len(negative_text) * 100, 1
        )
    })

theme_df = pd.DataFrame(theme_results)
display(
    theme_df.sort_values(
        "Matching reviews", ascending=False
    )
)
```

```python

# Visualize keyword-based complaint themes

import matplotlib.pyplot as plt
import seaborn as sns

plot_df = theme_df.sort_values(
    "Matching reviews",
    ascending=True
)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=plot_df,
    x="Matching reviews",
    y="Theme"
)

# Add counts at the end of each bar
for container in ax.containers:
    ax.bar_label(container, padding=4)

plt.title("Keyword-Based Themes in Negative Skincare Reviews")
plt.xlabel("Number of Matching Reviews")
plt.ylabel("Complaint Theme")
plt.tight_layout()

plt.savefig(
    "complaint_theme_analysis.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# Step: Compare sentiment across skincare products

product_sentiment = (
    sample_reviews
    .groupby(["brand_name", "product_name"])
    .agg(
        total_reviews=("sentiment", "count"),
        positive_reviews=(
            "sentiment",
            lambda x: (x == "Positive").sum()
        ),
        negative_reviews=(
            "sentiment",
            lambda x: (x == "Negative").sum()
        )
    )
    .reset_index()
)

product_sentiment["negative_percentage"] = (
    product_sentiment["negative_reviews"]
    / product_sentiment["total_reviews"] * 100
).round(1)

# Keep products with at least 5 reviews
product_sentiment = product_sentiment[
    product_sentiment["total_reviews"] >= 5
]

# Show products with the highest negative sentiment percentage
product_sentiment = product_sentiment.sort_values(
    "negative_percentage",
    ascending=False
)

display(product_sentiment.head(10))
```

```python

# Plot the top 10 products by negative sentiment percentage

import matplotlib.pyplot as plt
import seaborn as sns

top_products = product_sentiment.head(10).copy()

top_products["product_label"] = (
    top_products["brand_name"].astype(str)
    + " — "
    + top_products["product_name"].astype(str)
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_products.sort_values("negative_percentage"),
    x="negative_percentage",
    y="product_label"
)

plt.title("Products with the Highest Negative Sentiment")
plt.xlabel("Negative Sentiment (%)")
plt.ylabel("Brand and Product")
plt.tight_layout()

plt.savefig(
    "top_products_negative_sentiment.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
